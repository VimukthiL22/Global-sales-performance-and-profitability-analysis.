```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Vimukthi Lakshan\Desktop\My Projects\sales data\global_ecommerce_sales.csv")
```

```python
df
```

```python
# 1. Create the Product Catalog (Dimension Table)
# We drop duplicates so each product exists only once
product_catalog = df[['Product_Name', 'Product_Category']].drop_duplicates().reset_index(drop=True)
```

```python
product_catalog
```

```python

# We remove the descriptive columns that are now in the catalog, 
# keeping 'Product_Name' as the link (Foreign Key).
sales_data = df.drop(columns=['Product_Category', 'Unit_Price'])
```

```python
sales_data
```

```python
# Save to new CSV files
product_catalog.to_csv('product_catalog.csv', index=False)
sales_data.to_csv('sales_data.csv', index=False)

print("Tables successfully split!")
```
